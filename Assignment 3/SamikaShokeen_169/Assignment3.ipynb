```python
##  count digits in number
def count_num(n,c=0):
  while n>0:
    c+=1
    n=n//10
  return c
print(count_num(1895457))
```

```python
## check palindrome
def palindrome(n):
  l=0
  r=len(n)-1
  while l<r:
    if n[l]!=n[r]:
      return(False)
      break
    else:
      l+=1
      r-=1
  return(True)
n=input("enter number")
print(palindrome(n))
```

```python
##Fibonacci
def fibonacci(n):
  a=0
  b=0
  c=1
  while a<n:
    a=b
    b=c
    c=a+b
    print (c)
(fibonacci(10))
```
